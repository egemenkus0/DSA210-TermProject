# Hypothesis Testing
Significance level: α = 0.05

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, ttest_ind
# setting the environment
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
pd.options.display.float_format = '{:.2f}'.format

# creating the datasets.
burnout = pd.read_csv('burnout_processed.csv')
health = pd.read_csv('health_processed.csv', parse_dates=['date'])

# the comparison groups
exam = health[health['is_exam_period'] == 1]
nonexam = health[health['is_exam_period'] == 0]
```

## 1. Correlation Tests (Burnout Dataset)

```python
def hypothesis_test(test_name, var_name, target_name, df):
    
    # isolating the two variables and remove rows with missing data
    subset = df[[var_name, target_name]].dropna()
    
    # calculate pearson-r and significance (p-val)
    corr, p_val = pearsonr(subset[var_name], subset[target_name])
    
    # determining the direction of the relationship
    if corr > 0:
        direction = "POSITIVE"
    else:
        direction = "NEGATIVE"

    # printing the results.
    print(f'HYPOTHESIS TEST: {test_name}')
    print(f'   H0: No correlation between {var_name} and {target_name} (r = 0).')
    print(f'   H1: Significant correlation exists (r != 0).')
    print(f'   Pearson r: {corr:.4f} ({direction}),  P-Value: {p_val:.6f}')
    if p_val < 0.05:
        print(f'   *** REJECT H0 — significant relationship exists.\n')
    else:
        print(f'   *** FAIL TO REJECT H0 - no correlation found.\n')

hypothesis_test('Physical Activity & Burnout', 'physical_activity', 'burnout_score', burnout)
hypothesis_test('Sleep Hours & Burnout', 'sleep_hours', 'burnout_score', burnout)
hypothesis_test('Stress Level & Burnout', 'stress_level', 'burnout_score', burnout)
hypothesis_test('Social Support & Burnout', 'social_support', 'burnout_score', burnout)
```

## 2. t-Test — Exam vs Non-Exam (Apple Health Data)

```python
def ttest_hypothesis(test_name, col, exam_df, nonexam_df):
    # extract data for both groups and remove missing values
    exam_values = exam_df[col].dropna()
    nonexam_values = nonexam_df[col].dropna()

    # t_test required variables.
    t, p_val = ttest_ind(exam_values, nonexam_values, equal_var=False)

    # printing the result
    print(f'HYPOTHESIS TEST: { test_name }')
    print(f'   H0: No significant difference between exam and non-exam periods.')
    print(f'   H1: There IS a significant difference.')
    print(f'   Exam mean: {exam_values.mean():.2f}  |  Non-Exam mean: {nonexam_values.mean():.2f} ')
    print(f'   t-statistic: {t:.4f},  P-Value: {p_val:.6f}')
    if p_val < 0.05:
        print(f'   *** REJECT H0 — exam periods significantly affect { col }.\n')
    else:
        print(f'   *** FAIL TO REJECT H0 — no significant difference found.\n')

ttest_hypothesis('Daily Steps: Exam vs Non-Exam', 'daily_steps', exam, nonexam)
ttest_hypothesis('Exercise Minutes: Exam vs Non-Exam', 'exercise_minutes', exam, nonexam)
```
